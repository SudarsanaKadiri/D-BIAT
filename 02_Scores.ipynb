# 02 · Participant-level scores and D-scores; task version by group

**Scores** (from the retained trials of notebook 01; `dbiat_analysis.subject_measures`):

| Measure | Definition |
|---|---|
| Life:Me RT, Death:Me RT (overall) | mean RT of all retained trials of that block type |
| Life:Me RT, Death:Me RT (correct) | mean RT of the correct retained trials |
| Life:Me ER, Death:Me ER | 1 − proportion correct |
| D_RT | (mean RT Life:Me − mean RT Death:Me) / SD of all retained RTs of the participant |
| D_ER | ER Life:Me − ER Death:Me |
| D_Composite | mean of (RT / participant's mean RT + error) over Life:Me trials − the same over Death:Me trials |

Negative D-scores indicate faster or more accurate responding in Life:Me than in Death:Me blocks.

```python
import sys
sys.path.insert(0, "..")          # dbiat_analysis.py is in the folder above notebooks/
import pandas as pd
import dbiat_analysis as A

cfg = A.load_settings()
S = cfg["stats"]
clean = pd.read_csv(A.path(cfg, "derived", "trials_clean.csv"))
```

## 1. Scores

```python
scores = A.subject_measures(clean)
assert len(scores) == 135
scores.to_csv(A.path(cfg, "derived", "scores.csv"), index=False)
scores.groupby("group")[A.MEASURES].mean().reindex(A.GROUPS).round(4)
```

## 2. Task version by group (Methods, Task implementations)
χ² test with a Monte Carlo p-value (20,000 permutations).

```python
tab = scores.groupby(["group", "task_version"]).size().unstack().reindex(A.GROUPS)
chi2, p = A.perm_chi2(tab.values, S["n_chi2_permutations"], S["random_seed"])
print(f"Task version by group: χ²(2) = {chi2:.2f}, Monte Carlo {A.p_eq(p)}")
A.write_table(tab.reset_index().assign(chi2=chi2, p_monte_carlo=p), cfg, "task_version_by_group")
tab
```
